```python
import MDAnalysis as mda
u = mda.Universe("/Users/alice/Desktop/LNP project/trollman/trajectories/periodic_patches/low_ph/single_patch/low_ph_small_patch_replica_1.pdb")
print(u)
print("residue types:", set(u.residues.resnames))
```

```python
lipids = ["ACLP", "DSPC", "CHL1", "ACL"]
counts = []
for name in lipids:
    n = u.select_atoms("resname " + name).n_residues
    counts.append(n)
    print(name, n)
```

```python
table = pd.DataFrame({"lipid": lipids, "count": counts})
table["my %"] = table["count"] / table["count"].sum() * 100
table["published %"] = [46.3, 9.4, 42.7, 1.6]
table.round(1)
```
